Structure quantity - PV Riodades
=
This notebook counts the continuous structure rows in a list of structure positions and writes an AutoCAD script that draws a circle at the first (westernmost) position of each one, so the count can be checked on the drawing.

**Workflow overview:**
1. Read the structure positions from Excel.
2. Sort them row by row (Y descending) and west to east (X ascending).
3. Flag a position as `First` when it starts a new row or when there is a gap of more than 20 m to the previous one. The number of flags is the structure count.
4. Write a `CIRCLE` (radius 5) at every `First` position to `circles.scr`.

**Input prerequisites:**
- **File:** `C:\Users\Usuario\Desktop\Estruturas Riodades.xlsx`, first sheet.
- **Columns:** `Position X`, `Position Y` (used), and `EF`, `INV`, `Layer`, `PST`, `UFV` (dropped by name, so they must exist or the drop fails). Exact, case-sensitive names.
- **Positions:** one row per position along the structures. Positions of the same row have the same Y (to 1 decimal), and consecutive positions of one structure are at most 20 m apart in X (the pitch here is ~19.8 m).
- **No empty cells** in the kept columns: the `dropna()` of the script cell drops any row with a NaN.

Counting the structures
=
- Drops the tag columns and sorts by `Position Y` descending, then `Position X` ascending.
- `First` is `True` when:
  1. `Position Y` (rounded to 1 decimal) differs from the previous row (new row), or
  2. the previous X + **20 m** is smaller than this X (both rounded to 1 decimal): a gap, so a new structure.
- `contador` = number of `True` flags = number of structures (542 in the saved output).
- The 20 m threshold is close to the ~19.8 m pitch, so a small drawing error between two positions of the same structure is enough to split it.

```python
import pandas as pd
import numpy as np

estruturas = pd.read_excel(r"C:\Users\Usuario\Desktop\Estruturas Riodades.xlsx")
estruturas = estruturas.drop(columns=["EF", "INV", "Layer", "PST", "UFV"])
estruturas = estruturas.sort_values(by=['Position Y', 'Position X'], ascending=[False, True])

estruturas['First'] = np.where(
    (estruturas['Position Y'].round(1) != estruturas['Position Y'].shift().round(1)) | (estruturas['Position X'].shift().round(1) + 20 < estruturas['Position X'].round(1)),
    True,
    False
)

contador = estruturas['First'].sum()
print(contador)
print(estruturas)
```

Marking the first position of each structure
=
- `Script` = `CIRCLE <X>,<Y> 5` for the `First` rows, and NaN for the others. `dropna()` then keeps only the circles.
- Writes the lines to `C:\Users\Usuario\Desktop\circles.scr` (hardcoded path, overwritten). Running it in AutoCAD draws a circle of radius 5 on the current layer at the start of every structure.

```python
estruturas['Script'] = np.where(
    estruturas['First'] == True,
    "CIRCLE " + estruturas['Position X'].astype(str) + "," + estruturas['Position Y'].astype(str) + " 5",
    np.nan
)
estruturas = estruturas.dropna()
print(estruturas)

with open(r"C:\Users\Usuario\Desktop\circles.scr", 'w') as f:
    for line in estruturas['Script']:
        f.write(line)
        f.write("\n")
```
