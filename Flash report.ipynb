```python
import pandas as pd
import numpy as np


modules_received = pd.read_excel(r"C:\Users\Usuario\Desktop\Pessoal\Python\Ecotechne\Flash report\Consolidado 18.09.25.xlsx", names=['Nome da Origem', "No.", "Model Type", "S/N", "Isc(A)", "Voc(V)", "Pmax(W)", "Vpm(V)", "Ipm(A)", "FF", "Pallet No.", "ColorGrade"])

print(modules_received)
```

```python
column_name_patterns = ['Nome da Origem', "No.", "Model Type", "S/N", "Isc(A)", "Voc(V)", "Pmax(W)", "Vpm(V)", "Ipm(A)", "FF", "Pallet No.", "ColorGrade"]
mask = modules_received.apply(lambda row: not any(str(cell) in column_name_patterns for cell in row), axis=1)
modules = modules_received[mask].reset_index(drop=True)
print(modules)
```

```python
print(modules.describe(include='all'))
```

```python
print(modules.info())
```

```python
columns_str_to_numeric = ["Isc(A)", "Voc(V)", "Pmax(W)", "Vpm(V)", "Ipm(A)", "ColorGrade"]
for column in columns_str_to_numeric:
        modules[column] = pd.to_numeric(modules[column].str.replace(',', '.'), errors='coerce')


modules["FF"] = pd.to_numeric(modules["FF"].str[:-1])/100

print(modules.info())
print(modules.describe())
```

```python
modules = modules[modules["Ipm(A)"] > 16.5]

print(modules.info())
print(modules.describe(include="all"))
```

```python
modules["Container No."] = modules["Nome da Origem"].str.extract(r'^(?:[^_]*_){3}([^_]*)_')
modules = modules.drop(["Nome da Origem", "S/N", "ColorGrade", "Isc(A)", "Voc(V)"], axis=1)

modules['Pnom(W)'] = pd.to_numeric(modules["Model Type"].astype(str).str[4:7])

print(modules)
```

```python
modules = modules.sort_values(by=["Container No.", "Pallet No.", "Ipm(A)"]).reset_index(drop=True)
print(modules)
```

```python
modules["Invalid"] = np.where(
    (modules["Pmax(W)"] < modules["Pnom(W)"]) | (modules["Ipm(A)"] > 17.82),
    True,
    False
)



grouped_by_pallet = modules.groupby(by="Pallet No.", sort=False)
pnom_consistency = grouped_by_pallet['Pnom(W)'].nunique() == 1
print(pnom_consistency.value_counts())
print(grouped_by_pallet["Pnom(W)"].value_counts())

min_per_pallet = grouped_by_pallet['Ipm(A)'].min()
max_per_pallet = grouped_by_pallet['Ipm(A)'].max()

deviation = max_per_pallet - min_per_pallet
for index, value in deviation.items():
    if value > 0.6:
        print(index, value)
```

```python
all_pallets = modules.sort_values(by='Pnom(W)')['Pallet No.'].unique()

pallets_to_group = {}
for i, pallet in enumerate(all_pallets):
    group_num = i//82
    pallets_to_group[pallet] = group_num

modules['Pallet Group'] = modules['Pallet No.'].map(pallets_to_group)

grouped_by_82_pallets = modules.groupby(by="Pallet Group")
pnom_consistency2 = grouped_by_82_pallets["Pnom(W)"].nunique() == 1
print(pnom_consistency2.value_counts())

min_per_group = grouped_by_82_pallets['Ipm(A)'].min()
max_per_group = grouped_by_82_pallets['Ipm(A)'].max()

deviation = max_per_group - min_per_group
for index, value in deviation.items():
    if value > 0.6:
        print(index, value)

print(deviation[deviation > 0.4].count())
print(deviation.count())
```

```python
valid_data = modules[modules["Invalid"] == False]

modules.loc[valid_data.index, "Classification of Current Rating"], bins = pd.qcut(valid_data["Ipm(A)"], 3, labels=["High", "Medium", "Low"], retbins=True)
print(modules['Classification of Current Rating'].value_counts())
print("Cut points (bin edges):", bins)
print(f"Tercile ranges:")
print(f"Low: {bins[0]:.4f} to {bins[1]:.4f}")
print(f"Medium: {bins[1]:.4f} to {bins[2]:.4f}")
print(f"High: {bins[2]:.4f} to {bins[3]:.4f}")
```

```python
from collections import defaultdict, deque

class ModuleDistributor:
    def __init__(self, data):
        self.data = data
        self.pallet_groups = self._group_by_pallet()

    def _group_by_pallet(self):
        """Group modules by their pallet ID"""
        return self.data.groupby('Pallet No.')

    def get_pallet_sizes(self):
        """Get the number of modules in each pallet"""
        return self.pallet_groups.size()

    def distribute_strings(self):
        """
        Distribute modules into strings of 30, keeping pallets together
        Returns list of strings (each string is a list of module IDs)
        """
        # Sort pallets by size (descending) for better packing
        pallet_sizes = self.get_pallet_sizes()
        sorted_pallets = pallet_sizes.sort_values(ascending=False).index.tolist()

        strings = []
        current_string = []
        current_count = 0

        for pallet_id in sorted_pallets:
            pallet_modules = self.pallet_groups.get_group(pallet_id)
            pallet_size = len(pallet_modules)

            # If adding this pallet would exceed 30, start new string
            if current_count + pallet_size > 30:
                if current_string:  # Save current string if not empty
                    strings.append(current_string)

                # Start new string with this pallet
                current_string = pallet_modules.index.tolist()
                current_count = pallet_size

                # If this pallet alone exceeds 30, we have a problem
                if pallet_size > 30:
                    print(f"Warning: Pallet {pallet_id} has {pallet_size} modules (>30)")

            else:
                # Add pallet to current string
                current_string.extend(pallet_modules.index.tolist())
                current_count += pallet_size

                # If we reached exactly 30, start new string
                if current_count == 30:
                    strings.append(current_string)
                    current_string = []
                    current_count = 0

        # Add the last string if it has modules
        if current_string:
            strings.append(current_string)

        return strings

    def analyze_strings(self, strings):
        """Analyze the quality of the string distribution"""
        analysis = []

        for i, string in enumerate(strings):
            string_modules = self.data[self.data.index.isin(string)]

            analysis.append({
                'string_id': i + 1,
                'module_count': len(string),
                'avg_current': string_modules['Ipm(A)'].mean(),
                'current_std': string_modules['Ipm(A)'].std(),
                'total_power': string_modules['Pmax(W)'].sum(),
                'power_distribution': string_modules['Pmax(W)'].value_counts().to_dict(),
                'pallet_count': string_modules['Pallet No.'].nunique(),
                'module_ids': string
            })

        return pd.DataFrame(analysis)


# Main execution
if __name__ == "__main__":
    # Load your data (replace this with your actual data loading)
    # df = pd.read_csv('your_data.csv')


    # Initialize distributor
    distributor = ModuleDistributor(modules)

    # Get pallet sizes distribution
    pallet_sizes = distributor.get_pallet_sizes()
    print("\nPallet size distribution:")
    print(pallet_sizes.value_counts().sort_index())

    # Distribute into strings
    strings = distributor.distribute_strings()

    # Analyze the strings
    analysis = distributor.analyze_strings(strings)

    print(f"\nCreated {len(strings)} strings:")
    print(analysis[['string_id', 'module_count', 'avg_current', 'total_power', 'pallet_count']])

    # Check for strings that aren't exactly 30 modules
    non_30_strings = analysis[analysis['module_count'] != 30]
    if len(non_30_strings) > 0:
        print(f"\nStrings not exactly 30 modules: {len(non_30_strings)}")
        print(non_30_strings[['string_id', 'module_count']])

    # Save results to files
    analysis.to_csv('string_analysis.csv', index=False)

    # Save module-to-string mapping
    string_mapping = []
    for i, string in enumerate(strings):
        for module_id in string:
            string_mapping.append({
                'module_id': module_id,
                'string_id': i + 1
            })

    pd.DataFrame(string_mapping).to_csv('module_string_mapping.csv', index=False)
    print("\nResults saved to 'string_analysis.csv' and 'module_string_mapping.csv'")
```
