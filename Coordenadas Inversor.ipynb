```python
import pandas as pd
import numpy as np

df_INV = pd.read_excel(r"C:\Users\Usuario\Desktop\Inversores Sendim.xls")
df_EST = pd.read_excel(r"C:\Users\Usuario\Desktop\Estruturas Sendim.xls")

df_INV = df_INV.sort_values(by= "GRP").reset_index(drop=True)
df_INV = df_INV.drop(columns=["DSJ", "PVM", "QBT", "VBT", "INV", "PST", "UFV"])
df_INV = df_INV.rename(columns= {"Position X": "Inverter X", "Position Y": "Inverter Y"})

df_EST = df_EST.sort_values(by= ["UFV", "PST", "INV", "STR"]).reset_index(drop=True)
df_EST["GRP"] = df_EST["UFV"].astype(str)+"."+df_EST["PST"].astype(str)+"."+df_EST["INV"].astype(str).str.zfill(2)

struct_x = df_EST["Position X"].values
struct_y = df_EST["Position Y"].values
inv_x = df_INV["Inverter X"].values
inv_y = df_INV["Inverter Y"].values

x_dist = np.abs(struct_x[:, None] - inv_x[None, :])
y_dist = np.abs(struct_y[:, None] - inv_y[None, :])

within_x = x_dist <= 20
within_y = y_dist <= 2.3
within_both = within_x & within_y

results_df = pd.DataFrame(
    within_both,
    index = df_EST.index,
    columns = df_INV.index
).rename_axis("Structure").rename_axis("Inverter", axis=1)

true_counts = results_df.sum(axis=0)
print(true_counts.sum())

print(true_counts)
print(results_df)
```
