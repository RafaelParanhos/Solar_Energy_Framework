LV Trench Quantities per PST - Sendim
=
This notebook totals the LV trench length of PV Sendim per PST and trench layer and saves the table to Excel.

**Workflow overview:**
1. Read the LV network table.
2. Remove the rows with zeros or empty cells, and the circuit count column.
3. Sum the length per `PST` and `Layer` and export it.

**Input prerequisites (`PV Sendim - Rede de Baixa Tensão.xlsx`, on the Desktop)** (the code assumes all of these, and it doesn't check them):
- Columns `PST`, `Layer`, `Length` and `Número de circuitos` (this one is dropped, so it must exist). Any other numeric column is summed too.
- No `0` or empty cell in **any** column of a row that should be counted (see the warning below).

Summing the trench length per PST
=
- `replace(0, np.nan).dropna()` removes every row that has a `0` or an empty cell in **any** column, including `Número de circuitos`, which is dropped right after. A trench with `Número de circuitos = 0`, or with any other empty field, is left out of the totals.
- `groupby(['PST', 'Layer']).sum()` gives the total `Length` per PST and layer. The layer name already contains the PST (`..._PST_01_Plot_02`), so `PST` mostly makes the table easier to read.
- Output: **Valas por PST.xlsx** on the Desktop.

```python
import pandas as pd
import numpy as np

valas = pd.read_excel(r"C:\Users\Usuario\Desktop\PV Sendim - Rede de Baixa Tensão.xlsx")

valas = valas.replace(0,np.nan).dropna()
valas = valas.drop(columns="Número de circuitos")

valas_by_pst = valas.groupby(['PST', 'Layer']).sum()

print(valas_by_pst)

valas_by_pst.to_excel(r"C:\Users\Usuario\Desktop\Valas por PST.xlsx")
```
