```python
import pandas as pd
import glob
import os

folder_path = r"C:\Users\Usuario\Ecotechnee\Office - Documentos\Projetos\309 - Zagope - UVF Riodades\02 - DR\03 - Equipamentos\01 - Módulo\01 - Flash Reports\02 - 04.03.26"
# Get a list of CSV file paths
file_paths = glob.glob(os.path.join(folder_path, r'*\*.xls'))

# Create a list of DataFrames, one for each file
df_list = [pd.read_excel(f) for f in file_paths]
print(df_list)

# Concatenate all DataFrames into a single one
modules_flash_report = pd.concat(df_list, ignore_index=True)

df_columns = ["No.", "Model Type", "Module No.", "Isc(A)", "Voc(V)", "Pmax(W)", "Vpm(V)", "Ipm(A)", "FF", "Pallet No.", "Color Grade"]

modules_flash_report.columns = df_columns
```

```python
modules_flash_report.info()
```

```python
modules_flash_report.dropna(how="any", inplace=True)
modules_flash_report.drop_duplicates(subset="Module No.", inplace=True, ignore_index=True)
modules_flash_report.drop(index=0, inplace=True)
modules_flash_report.info()
print(modules_flash_report.head())
```

```python
modules_flash_report["Power"] = modules_flash_report["Model Type"].astype(str).str[4:7].astype(int)
modules_flash_report["Module Intensity"] = modules_flash_report['Pallet No.'].astype(str).str[-1].astype(int)
modules_flash_report['Power - Intensity'] = modules_flash_report["Power"].astype(str) + "-" + modules_flash_report["Module Intensity"].astype(str)
print(modules_flash_report.head())
```

```python
power_intensity_groups = modules_flash_report.groupby(by="Power - Intensity")
print(power_intensity_groups.size())
```

```python
power_groups = modules_flash_report.groupby(by="Power")
print(power_groups.size())
```
