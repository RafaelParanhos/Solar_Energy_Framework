```python
import pandas as pd
import numpy as np


df_estruturas = pd.read_excel(r"C:\Users\Usuario\Desktop\Pessoal\Python\Ecotechne\Separador de Strings\Estruturas Capela.xlsx")
df_inversores = pd.read_excel(r"C:\Users\Usuario\Desktop\Pessoal\Python\Ecotechne\Separador de Strings\Inversores Capela.xlsx")


print(df_estruturas)
print(df_inversores)
```

```python
df_inversores["GRP"] = df_inversores["PRK"].astype(str) + "." + df_inversores["UFV"].astype(str) + "." + df_inversores["PST"].astype(str)
df_estruturas["GRP"] = df_estruturas["PRK"].astype(str) + "." + df_estruturas["UFV"].astype(str) + "." + df_estruturas["PST"].astype(str)

df_estruturas['TAG'] = df_estruturas['GRP'].astype(str) + '.' + df_estruturas['INV'].astype(str).str.zfill(2)
df_inversores['TAG'] = df_inversores['GRP'].astype(str) + '.' + df_inversores['INV'].astype('Int64').astype(str).str.zfill(2)

print(df_estruturas)
print(df_inversores)
```

```python
df_estruturas['Comparing X'] = df_estruturas.groupby('TAG')['Position X'].transform('mean')

df_estruturas = df_estruturas.merge(df_inversores, on='TAG')

df_estruturas['E or W'] = np.where(
    df_estruturas['Position X_y'] >= df_estruturas['Comparing X'],
    'E',
    'W'
)

df_estruturas.sort_values(by=['PRK_x', 'UFV_x', 'PST_x', 'INV_x', 'Position Y_x', 'Position X_x'], ascending=[True, True, True, True, False, True], inplace=True)

df_estruturas["STR"] = df_estruturas.groupby("TAG").cumcount()+1
df_estruturas['GRP2'] = df_estruturas['TAG']
df_estruturas["TAG"] = df_estruturas['TAG'] + "." + df_estruturas['STR'].astype(str).str.zfill(2)


print(df_estruturas)
print(df_estruturas["TAG"])
```

```python
def group_structures(df, inverter_position, max_distance=22):
    """
    Groups adjacent structures in pairs (2) while respecting:
    - TAG boundaries
    - Maximum distance constraint (22m default)
    - Sorted by Position Y then X
    """
    # Sort by Position X (primary) and Position Y (secondary)

    df = df.sort_values(['Position Y_x', 'Position X_x'], ascending=[False, True]).reset_index(drop=True)



    # Calculate pairwise distances between consecutive structures
    df['x_diff'] = df['Position X_x'].diff()
    df['y_diff'] = df['Position Y_x'].diff()
    df['distance'] = np.sqrt(df['x_diff']**2 + df['y_diff']**2)

    # Create group boundaries where:
    # 1. TAG changes OR #### THIS IS NO LONGER A BOUNDARY ####
    # 2. Distance > max_distance OR
    # 3. It's the first structure
    df['group_boundary'] = (
        #(df['TAG'] != df['TAG'].shift(1)) |
        (df['distance'] > max_distance) |
        (df.index == 0)
    )
    # Create group IDs
    df['group_id'] = df['group_boundary'].cumsum()

    # Split into groups
    result_groups = []

    for _, group_df in df.groupby('group_id'):
        if inverter_position == "W":
            group_df = group_df.iloc[::-1].reset_index(drop=True)
        # Split groups larger than 2 into pairs
        size = len(group_df)
        if size <= 2:
            result_groups.append(group_df)

            continue

        n_pairs = size // 2
        for i in range(n_pairs):
            result_groups.append(group_df.iloc[i*2:i*2+2])


        if size % 2:  # Handle odd-sized groups
            result_groups.append(group_df.iloc[-1:])

    if inverter_position == "W":
        result_groups = result_groups[::-1]

    return result_groups

def create_group_positions(groups, inverter_position):
    """Create the output DataFrame from groups (same as before)"""
    group_data = []
    for i, group_df in enumerate(groups, 1):
        if len(group_df) > 1:
            if inverter_position == "W":
                first_struct = group_df.iloc[1]
                second_struct = group_df.iloc[0]
            else:
                first_struct = group_df.iloc[0]
                second_struct = group_df.iloc[1]
            group_data.append({
                'Group_ID': i,
                'TAG-1': first_struct['TAG'],
                'TAG-2': second_struct['TAG'],
                'Position_X': first_struct['Position X_x'],
                'Position_Y': first_struct['Position Y_x'],
                'Group_Size': len(group_df),
                'Layer-1': first_struct['Layer_x'],
                'Layer-2': second_struct['Layer_x']
            })
        else:
            first_struct = group_df.iloc[0]
            group_data.append({
                'Group_ID': i,
                'TAG-1': first_struct['TAG'],
                'Position_X': first_struct['Position X_x'],
                'Position_Y': first_struct['Position Y_x'],
                'Group_Size': len(group_df),
                'Layer-1': first_struct['Layer_x']
            })
    return pd.DataFrame(group_data)
```

```python
estruturas_E = df_estruturas[df_estruturas['E or W'] == 'E']
estruturas_W = df_estruturas[df_estruturas['E or W'] == 'W']


results_E = group_structures(estruturas_E, 'E')
results_W = group_structures(estruturas_W, 'W')

group_positions_E = create_group_positions(results_E, "E")
group_positions_W = create_group_positions(results_W, "W")

group_positions = pd.concat([group_positions_E, group_positions_W], ignore_index=True)

group_positions['PRK-1'] = group_positions['TAG-1'].str[0]
group_positions['PRK-2'] = np.where(
    group_positions["Group_Size"] == 2,
    group_positions['TAG-2'].str[0],
    None
)
group_positions["UFV-1"] = group_positions['TAG-1'].str[2]
group_positions['UFV-2'] = np.where(
    group_positions["Group_Size"] == 2,
    group_positions['TAG-2'].str[2],
    None
)
group_positions["PST-1"] = group_positions['TAG-1'].str[4]
group_positions['PST-2'] = np.where(
    group_positions["Group_Size"] == 2,
    group_positions['TAG-2'].str[4],
    None
)
group_positions["INV-1"] = group_positions['TAG-1'].str[6:8]
group_positions['INV-2'] = np.where(
    group_positions["Group_Size"] == 2,
    group_positions['TAG-2'].str[6:8],
    None
)
group_positions["STR-1"] = group_positions['TAG-1'].str[-2:]
group_positions['STR-2'] = np.where(
    group_positions["Group_Size"] == 2,
    group_positions['TAG-2'].str[-2:],
    None
)
group_positions['GRP-1'] = group_positions['TAG-1'].str[:-6]
group_positions['GRP-2'] = np.where(
    group_positions["Group_Size"] == 2,
    group_positions['TAG-2'].str[:-6],
    None
)

correction_position_x = 10.684
correction_position_y = 1.1613

# For row-wise operation (recommended)
group_positions['Script'] = np.where(
    group_positions['Group_Size'] == 1,
    group_positions.apply(
        lambda row: (
            f'(command-s "._layer" "set" "{row["Layer-1"]}" "")\n'
            f'(command-s "._insert" "EMF379_string_{row["Group_Size"]}" '
            f'"{row["Position_X"]},{row["Position_Y"]}" 1 1 0 {row["PRK-1"]} {row["UFV-1"]} {row["PST-1"]} {row["INV-1"]} "" "{row["STR-1"]}" "{row["GRP-1"]}" 730)'
        ),
    axis=1
    ),
    np.where(
        group_positions["Layer-1"] == group_positions["Layer-2"],
        group_positions.apply(
            lambda row: (
                f'(command-s "._layer" "set" "{row["Layer-1"]}" "")\n'
                f'(command-s "._insert" "EMF379_string_{row["Group_Size"]}" '
                f'"{round(row["Position_X"]+correction_position_x,4)},{round(row["Position_Y"]+correction_position_y,4)}" 1 0 {row["PRK-1"]} {row["UFV-1"]} {row["PST-1"]} {row["INV-1"]} "" "{row["STR-1"]}" "{row["GRP-1"]}" 730)\n'
                f'(command-s "._insert" "EMF379_string_{row["Group_Size"]}" '
                f'"{round(row["Position_X"]+correction_position_x,4)},{round(row["Position_Y"]-correction_position_y,4)}" 1 0 {row["PRK-2"]} {row["UFV-2"]} {row["PST-2"]} {row["INV-2"]} "" "{row["STR-2"]}" "{row["GRP-2"]}" 730)'
            ),
        axis=1
        ),
        group_positions.apply(
            lambda row: (
                f'(command-s "._layer" "set" "{row["Layer-1"]}" "")\n'
                f'(command-s "._insert" "EMF379_string_{row["Group_Size"]}" '
                f'"{round(row["Position_X"]+correction_position_x,4)},{round(row["Position_Y"]+correction_position_y,4)}" 1 0 {row["PRK-1"]} {row["UFV-1"]} {row["PST-1"]} {row["INV-1"]} "" "{row["STR-1"]}" "{row["GRP-1"]}" 730)\n'
                f'(command-s "._layer" "set" "{row["Layer-2"]}" "")\n'
                f'(command-s "._insert" "EMF379_string_{row["Group_Size"]}" '
                f'"{round(row["Position_X"]+correction_position_x,4)},{round(row["Position_Y"]-correction_position_y,4)}" 1 0 {row["PRK-2"]} {row["UFV-2"]} {row["PST-2"]} {row["INV-2"]} "" "{row["STR-2"]}" "{row["GRP-2"]}" 730)'
            ),
        axis=1
        )
    )
)
print(group_positions)

with open (r"C:\Users\Usuario\Desktop\Pessoal\Python\Ecotechne\Separador de Strings\Separador de Strings - Capela.scr", 'w') as f:
    f.write('\n'.join(group_positions["Script"].astype(str)) + '\n')

group_positions.to_excel(r"C:\Users\Usuario\Desktop\Pessoal\Python\Ecotechne\Separador de Strings\teste.xlsx")
```
