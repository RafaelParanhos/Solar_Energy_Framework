AC Trench Quantities by Number of Circuits
=
This notebook totals the length of the LV AC trenches (`valas BT CA`) by the number of circuits in each trench, which is read from the layer name.

**Workflow overview:**
1. Read the trench table exported from AutoCAD.
2. Take the number of circuits from the layer name.
3. Sum the trench length per number of circuits.

**Input prerequisites (`valas ca.xls`, in `...\Python\Qty Valas CA`)** (the code assumes all of these, and it doesn't check them):
- One row per trench polyline, with the columns `Layer` and `Length` (AutoCAD data extraction).
- Every layer follows the pattern `EMF_bt_vala_btca_NNxC...`, and the **19th character** (index 18) is the units digit of the circuit count (`EMF_bt_vala_btca_04xC` → `4`).
- Fewer than 10 circuits per trench: only one digit is read, so `10xC` would be counted as `0` and `11xC` as `1`.
- Only trench layers are in the table. Any other layer gives a meaningless key.

Summing the trench length per number of circuits
=
- `N Circuitos` = `Layer.str[18:19]`, a single character of the layer name (see the prerequisites).
- `groupby('N Circuitos').sum()` gives the total `Length` per circuit count. It also "sums" the `Layer` text, which just concatenates the layer names, so that column can be ignored.
- The result is only printed, it isn't saved.
- The `time` calls measure the run time. `math` and `numpy` are imported but not used, and `df_cabos = pd.DataFrame()` is overwritten on the next line.

```python
import time

import pandas as pd

import math

import numpy as np

start = time.time()

df_valas = pd.read_excel(r"C:\Users\Usuario\Desktop\Pessoal\Python\Qty Valas CA\valas ca.xls")

df_valas['N Circuitos'] = df_valas["Layer"].str[18:19]


df_cabos = pd.DataFrame()

df_cabos = df_valas.groupby('N Circuitos').sum()


print(df_cabos)


end = time.time()
elapsed = end - start
print(elapsed)
```
