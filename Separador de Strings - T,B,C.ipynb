Unfinished draft. It reads the Riodades structure list, takes the structure type from the last character of the block name (e.g. `EMF309_estrutura_fixa_2` → `2`), and sorts the structures top-to-bottom / left-to-right. It is meant to write a `CIRCLE` script marking the first structure of each group. That step doesn't run: it needs a `First` column that nothing creates.

**Input prerequisites** (the code assumes all of these, and it doesn't check them):
- **File:** `Estruturas Riodades.xlsx` in `C:\Users\Usuario\Desktop\Pessoal\Python\Separador de Strings CTB\` (hardcoded path).
- **Columns:** `Name`, `Position X`, `Position Y`, `INV`, `Layer`, `PST`, plus `EF` and `UFV` (they are dropped by name, so a missing one raises a `KeyError`).
- **Circle step:** also needs a boolean `First` column (not in the file and not computed here).

Loading and sorting the structures
=
- Reads the structure list and drops `EF` and `UFV`.
- `Type` = last character of `Name` (the structure size in the block name). It is not used afterwards.
- Sorts by `Position Y` descending, then `Position X` ascending (top-to-bottom, left-to-right).
- `numpy` is imported here for the next cell.

```python
import pandas as pd
import numpy as np

estruturas_strings = pd.read_excel(r"C:\Users\Usuario\Desktop\Pessoal\Python\Separador de Strings CTB\Estruturas Riodades.xlsx")
estruturas_strings = estruturas_strings.drop(columns=['EF', 'UFV'])

estruturas_strings['Type'] = estruturas_strings['Name'].astype(str).str[-1]
estruturas_strings = estruturas_strings.sort_values(by=['Position Y', 'Position X'], ascending=[False, True])


print(estruturas_strings)
```
