```python
import pandas as pd
import numpy as np

estruturas_strings = pd.read_excel(r"C:\Users\Usuario\Desktop\Pessoal\Python\Separador de Strings CTB\Estruturas Riodades.xlsx")
estruturas_strings = estruturas_strings.drop(columns=['EF', 'UFV'])

estruturas_strings['Type'] = estruturas_strings['Name'].astype(str).str[-1]
estruturas_strings = estruturas_strings.sort_values(by=['Position Y', 'Position X'], ascending=[False, True])


print(estruturas_strings)
```

```python
estruturas_strings['Script'] = np.where(
    estruturas_strings['First'] == True,
    "CIRCLE " + estruturas_strings['Position X'].astype(str) + "," + estruturas_strings['Position Y'].astype(str) + " 5",
    np.nan
)
test = estruturas_strings.dropna()

with open(r"C:\Users\Usuario\Desktop\circles.scr", 'w') as f:
    for line in test['Script']:
        f.write(line)
        f.write("\n")
```
